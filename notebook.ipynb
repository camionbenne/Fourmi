### Méthode d'Euler pour le calcul de solutions approchées

Pour cela on va rester avec la fonction  $y' = t - y²$, 

```python
def f(x:float,y:float)->float:
    """x - y²"""
    return x -y**2
```

```python
precision = 20 
subdivision = np.linspace(1,3,precision)
y_0, x_0 = 5,1
y = [y_0]
for i in range(len(subdivision)-1):
    y_n = y[-1]
    x_n = subdivision[i]
    step = subdivision[i+1] -subdivision [i]
    y.append(y_n +f(x_n,y_n)* step)

fig, ax = plt.subplots()
ax.plot(subdivision, y,"b:o")
ax.set_title("Application de la méthode d'Euler ")

plt.show()
```
